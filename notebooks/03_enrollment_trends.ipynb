# 3. Enrollment & Performance Trends
Analyze how enrollment and performance change over time (2022-2024).

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sys
sys.path.append('../scripts')
from helpers import load_education_data

df = load_education_data('../data/khi_education_demo.csv')
sns.set_style('whitegrid')
```

## Overall Trends: System-Wide Metrics

```python
# Year-over-year summary
yearly_summary = df.groupby('Year').agg({
    'Overall_Score': 'mean',
    'Enrollment': 'sum',
    'Attendance_Rate': 'mean',
    'School_Name': 'count'
}).round(2)
yearly_summary.rename(columns={'School_Name': 'Schools'}, inplace=True)

print("Year-over-Year Summary:")
print(yearly_summary)
```

```python
# Calculate year-over-year growth
print("\nYear-over-Year Change:")
print(f"Score Improvement (2022-2024): {(yearly_summary.loc[2024, 'Overall_Score'] - yearly_summary.loc[2022, 'Overall_Score']):.2f} points")
print(f"Enrollment Growth: {yearly_summary.loc[2024, 'Enrollment'] - yearly_summary.loc[2022, 'Enrollment']:.0f} students")
print(f"Attendance Improvement: {yearly_summary.loc[2024, 'Attendance_Rate'] - yearly_summary.loc[2022, 'Attendance_Rate']:.2f}%")
```

## Visualization: System-Wide Trends

```python
# Multi-metric trend visualization
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Overall Score Trend
axes[0, 0].plot(yearly_summary.index, yearly_summary['Overall_Score'], marker='o', linewidth=2, markersize=8, color='#1f77b4')
axes[0, 0].set_title('Average Overall Score Trend', fontsize=12, fontweight='bold')
axes[0, 0].set_ylabel('Score')
axes[0, 0].grid(True, alpha=0.3)

# Enrollment Trend
axes[0, 1].plot(yearly_summary.index, yearly_summary['Enrollment'], marker='s', linewidth=2, markersize=8, color='#ff7f0e')
axes[0, 1].set_title('Total Enrollment Trend', fontsize=12, fontweight='bold')
axes[0, 1].set_ylabel('Students')
axes[0, 1].grid(True, alpha=0.3)

# Attendance Trend
axes[1, 0].plot(yearly_summary.index, yearly_summary['Attendance_Rate'], marker='^', linewidth=2, markersize=8, color='#2ca02c')
axes[1, 0].set_title('Average Attendance Rate', fontsize=12, fontweight='bold')
axes[1, 0].set_ylabel('Attendance (%)')
axes[1, 0].grid(True, alpha=0.3)

# Growth comparison
growth_data = {
    'Score': ((yearly_summary.loc[2024, 'Overall_Score'] - yearly_summary.loc[2022, 'Overall_Score']) / yearly_summary.loc[2022, 'Overall_Score'] * 100),
    'Enrollment': ((yearly_summary.loc[2024, 'Enrollment'] - yearly_summary.loc[2022, 'Enrollment']) / yearly_summary.loc[2022, 'Enrollment'] * 100),
    'Attendance': ((yearly_summary.loc[2024, 'Attendance_Rate'] - yearly_summary.loc[2022, 'Attendance_Rate']) / yearly_summary.loc[2022, 'Attendance_Rate'] * 100)
}
axes[1, 1].bar(growth_data.keys(), growth_data.values(), color=['#1f77b4', '#ff7f0e', '#2ca02c'])
axes[1, 1].set_title('Growth 2022 to 2024 (%)', fontsize=12, fontweight='bold')
axes[1, 1].set_ylabel('Growth %')

plt.tight_layout()
plt.show()
```

## Individual School Trends

```python
# Select top and bottom performing schools
school_avg = df.groupby('School_Name')['Overall_Score'].mean().sort_values(ascending=False)
top_schools = school_avg.head(3).index.tolist()
bottom_schools = school_avg.tail(3).index.tolist()

print(f"Top 3 Schools: {top_schools}")
print(f"Bottom 3 Schools: {bottom_schools}")
```

```python
# Plot top schools trends
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Top performers
for school in top_schools:
    school_data = df[df['School_Name'] == school].sort_values('Year')
    axes[0].plot(school_data['Year'], school_data['Overall_Score'], marker='o', label=school, linewidth=2)
axes[0].set_title('Top 3 Schools: Performance Trend', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Year')
axes[0].set_ylabel('Overall Score')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Bottom performers
for school in bottom_schools:
    school_data = df[df['School_Name'] == school].sort_values('Year')
    axes[1].plot(school_data['Year'], school_data['Overall_Score'], marker='o', label=school, linewidth=2)
axes[1].set_title('Bottom 3 Schools: Performance Trend', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Year')
axes[1].set_ylabel('Overall Score')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## Regional Trends

```python
# Regional trends over time
regional_trends = df.groupby(['Year', 'Region']).agg({
    'Overall_Score': 'mean',
    'Enrollment': 'sum'
}).reset_index()

# Plot regional performance trends
fig, ax = plt.subplots(figsize=(12, 6))
for region in regional_trends['Region'].unique():
    region_data = regional_trends[regional_trends['Region'] == region].sort_values('Year')
    ax.plot(region_data['Year'], region_data['Overall_Score'], marker='o', label=region, linewidth=2)

ax.set_title('Regional Performance Trends', fontsize=12, fontweight='bold')
ax.set_xlabel('Year')
ax.set_ylabel('Average Overall Score')
ax.legend(loc='best')
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
