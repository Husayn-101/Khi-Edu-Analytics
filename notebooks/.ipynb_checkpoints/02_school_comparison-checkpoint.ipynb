# 2. School Performance Comparison
Compare performance metrics across different schools, school types, and regions.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sys
sys.path.append('../scripts')
from helpers import load_education_data, get_school_performance_summary, get_school_type_comparison

df = load_education_data('../data/khi_education_demo.csv')
sns.set_style('whitegrid')
```

## School Performance Summary

```python
school_perf = get_school_performance_summary(df)
school_perf
```

```python
# Average performance by school (across all years)
school_avg = df.groupby('School_Name').agg({
    'Overall_Score': 'mean',
    'Attendance_Rate': 'mean',
    'School_Type': 'first',
    'Region': 'first'
}).round(2).sort_values('Overall_Score', ascending=False)

print("School Performance Rankings:")
print(school_avg)
```

## Visualization: School Performance Comparison

```python
# Bar chart of average scores by school
fig, ax = plt.subplots(figsize=(12, 6))
school_avg_sorted = school_avg.sort_values('Overall_Score')
ax.barh(school_avg_sorted.index, school_avg_sorted['Overall_Score'], color='steelblue')
ax.set_xlabel('Average Overall Score')
ax.set_title('School Performance Rankings (2022-2024)')
plt.tight_layout()
plt.show()
```

## Public vs Private Schools Comparison

```python
school_type_comp = get_school_type_comparison(df)
print("\nPublic vs Private Schools Comparison:")
print(school_type_comp)
```

```python
# Compare school types
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Scores by type
df.groupby('School_Type')['Overall_Score'].mean().plot(kind='bar', ax=axes[0], color=['#ff7f0e', '#2ca02c'])
axes[0].set_title('Average Score: Public vs Private')
axes[0].set_ylabel('Overall Score')
axes[0].set_xlabel('')

# Attendance by type
df.groupby('School_Type')['Attendance_Rate'].mean().plot(kind='bar', ax=axes[1], color=['#ff7f0e', '#2ca02c'])
axes[1].set_title('Average Attendance Rate: Public vs Private')
axes[1].set_ylabel('Attendance Rate (%)')
axes[1].set_xlabel('')

plt.tight_layout()
plt.show()
```

## Regional Comparison

```python
# Regional performance
regional_avg = df.groupby('Region').agg({
    'Overall_Score': 'mean',
    'Enrollment': 'sum',
    'Attendance_Rate': 'mean',
    'School_Name': 'count'
}).round(2)
regional_avg.rename(columns={'School_Name': 'School_Count'}, inplace=True)

print("\nRegional Performance Summary:")
print(regional_avg.sort_values('Overall_Score', ascending=False))
```

```python
# Regional comparison visualization
regional_sorted = regional_avg.sort_values('Overall_Score')
fig, ax = plt.subplots(figsize=(10, 5))
ax.barh(regional_sorted.index, regional_sorted['Overall_Score'])
ax.set_xlabel('Average Overall Score')
ax.set_title('Performance by Region')
plt.tight_layout()
plt.show()
```
